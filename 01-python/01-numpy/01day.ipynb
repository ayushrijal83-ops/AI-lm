```python
import numpy as np
```

```python
#lets learn how we can create first dimension array.
a = np.array([1,5,4], dtype="int32")
print(a)
```

```python
#lets learn how we can create two dimension array.
b = np.array([[2.0, 6.0, 7.0 ], [3.0, 7.0, 1.0]])
print(b)
```

```python
#lets learn how we can simply know which dimension array we have created.
a.ndim
b.ndim
```

```python
#lets learn how was the shape of our array /get shape
a.shape
```

```python
#lets learn how much memory our numpy arrays take up, 
#get the data type
a.dtype
b.dtype
```

```python
#get the size
a.itemsize
b.itemsize
```

```python
#get the total size
a.nbytes
b.nbytes
```
