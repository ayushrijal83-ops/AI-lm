```python
import numpy as np
```

```python
#lets learn how we can create first dimension array.
a = np.array([1,5,4], dtype="int32")
print(a)
```

```python
#lets learn how we can create two dimension array.
b = np.array([[2.0, 6.0, 7.0 ], [3.0, 7.0, 1.0]])
print(b)
```

```python
#lets learn how we can simply know which dimension array we have created.
a.ndim
b.ndim
```

```python
#lets learn how was the shape of our array /get shape
a.shape
```

```python
#lets learn how much memory our numpy arrays take up, 
#get the data type
a.dtype
b.dtype
```

```python
#get the size
a.itemsize
b.itemsize
```

```python
#get the total size
a.nbytes
b.nbytes
```

```python
a = np.array([[11,12,13,14,15,16], [21, 22, 23, 24, 25, 26]])
print(a)
```

```python
#getting specific elements, row, coulums, etc.

#1) getting specific element[r, c]
# the meaning of row and column is we simply count and put the exact number of row and column which elemet we want to get.

a[0, 4]
```

```python
# getting specific row
a[0, :]
# so what does it mean it simply means print only the first row and all column on that row.
```

```python
#gertting specific column
a[:, 2]
#it means w ar going to print all the rows but only the one column
```

```python
#Getting a little more fancy [startindex:endindex:stepsize]
a[0,1:6:2]

#this way we can acess elements.
```

```python
#lets learn how we can change the elements specifi element so for that we can simply do it by like this 

print(a[1,4])

#lets change this element 25 for that we do it like this way..

a[1,4] = 10 # what does it means it mean change the the element 25 to 10 

#lets test it even if it was changed or not

print(a[1,4])
```

```python
#lets learn how to do changing the elements of entire column up to we want 
#for that we do it like this way 

print(a[:, 3])

#now lets change it
a[:, 3] = [1, 2]
print(a)
```
